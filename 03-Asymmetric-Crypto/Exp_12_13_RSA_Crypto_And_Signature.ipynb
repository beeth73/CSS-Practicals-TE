```python
# Exp_12_13_RSA_Crypto_And_Signature.py

# 1. From-scratch Modular Inverse function (To find Private Key 'd')
def mod_inverse(e, phi):
    for d in range(1, phi):
        if (e * d) % phi == 1:
            return d
    return -1

# 2. Key Generation
# In reality, p and q are massive primes. We use small ones for the lab.
p = 61
q = 53
n = p * q
phi = (p - 1) * (q - 1)

# e must be coprime to phi
e = 17 

# d is the modular inverse of e mod phi
d = mod_inverse(e, phi)

print(f"Public Key: (e={e}, n={n})")
print(f"Private Key: (d={d}, n={n})\n")

# --- EXPERIMENT 12: RSA ENCRYPTION ---
message = 65  # We encrypt the ASCII number for 'A'

# Encryption: C = M^e mod n (Using built-in pow for efficiency)
cipher_text = pow(message, e, n)
print(f"--- EXP 12: RSA ENCRYPTION ---")
print(f"Original Message : {message}")
print(f"Encrypted Cipher : {cipher_text}")

# Decryption: M = C^d mod n
decrypted_msg = pow(cipher_text, d, n)
print(f"Decrypted Message: {decrypted_msg}\n")


# --- EXPERIMENT 13: RSA DIGITAL SIGNATURE ---
# Signature: S = M^d mod n (Signed with PRIVATE KEY)
signature = pow(message, d, n)
print(f"--- EXP 13: RSA DIGITAL SIGNATURE ---")
print(f"Document/Message : {message}")
print(f"Generated Signature: {signature}")

# Verification: M' = S^e mod n (Verified with PUBLIC KEY)
verified_msg = pow(signature, e, n)
print(f"Verified Document: {verified_msg}")
print(f"Is Signature Valid? {message == verified_msg}")
```
