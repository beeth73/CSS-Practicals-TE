```python
# Exp_14_Diffie_Hellman.py

print("--- DIFFIE-HELLMAN KEY EXCHANGE ---")

# 1. Publicly known variables (Agreed upon over insecure channel)
q = 353  # A prime number
alpha = 3  # A primitive root modulo q
print(f"Publicly Shared -> Prime (q): {q}, Primitive Root (alpha): {alpha}")

# 2. Alice's Side
X_A = 97  # Alice's Private Key (Secret)
# Alice calculates her Public Key (Y_A) to send to Bob: Y_A = alpha^X_A mod q
Y_A = pow(alpha, X_A, q)
print(f"Alice sends Public Key (Y_A): {Y_A}")

# 3. Bob's Side
X_B = 233  # Bob's Private Key (Secret)
# Bob calculates his Public Key (Y_B) to send to Alice: Y_B = alpha^X_B mod q
Y_B = pow(alpha, X_B, q)
print(f"Bob sends Public Key (Y_B): {Y_B}")

# 4. The Magic Exchange (Calculating the Shared Secret)
# Alice receives Y_B and mixes it with her private key
Secret_Alice = pow(Y_B, X_A, q)

# Bob receives Y_A and mixes it with his private key
Secret_Bob = pow(Y_A, X_B, q)

print(f"\nAlice calculates Shared Secret: {Secret_Alice}")
print(f"Bob calculates Shared Secret:   {Secret_Bob}")
print(f"Success? {Secret_Alice == Secret_Bob}")
```
