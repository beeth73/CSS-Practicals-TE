```python
# Exp_18_Totient_Function.py

# Helper function to find Greatest Common Divisor (Euclidean Algorithm)
# We write it from scratch to avoid importing the math module!
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

# Helper function to check if a number is prime
def is_prime(num):
    if num <= 1: return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0: return False
    return True

def eulers_totient_rules(m, n):
    print("--- EULER'S TOTIENT FUNCTION RULES ---")
    
    # Rule 1: Φ(1) = 0
    print(f"Rule 1: Φ(1) = 0")
    
    # Rule 2: Φ(p) = p - 1 if p is prime.
    if is_prime(m):
        print(f"Rule 2: {m} is prime. Therefore, Φ({m}) = {m} - 1 = {m - 1}")
    else:
        print(f"Rule 2: {m} is NOT prime, rule doesn't apply directly.")
        
    # Rule 3: Φ(m * n) = Φ(m) * Φ(n) if m and n are relatively prime.
    if gcd(m, n) == 1:
        # Calculate Φ manually by counting coprimes
        phi_m = sum(1 for i in range(1, m) if gcd(m, i) == 1)
        phi_n = sum(1 for i in range(1, n) if gcd(n, i) == 1)
        phi_mn = sum(1 for i in range(1, m * n) if gcd(m * n, i) == 1)
        
        print(f"Rule 3: {m} and {n} are relatively prime (coprime).")
        print(f"Φ({m}) = {phi_m}")
        print(f"Φ({n}) = {phi_n}")
        print(f"Φ({m} * {n}) = Φ({m*n}) = {phi_mn}")
        print(f"Proof: {phi_m} * {phi_n} = {phi_m * phi_n} == {phi_mn}")
    else:
        print(f"Rule 3 fails: {m} and {n} share a common factor.")

# Run the test with a prime (7) and a non-prime coprime (10)
eulers_totient_rules(7, 10)
```
