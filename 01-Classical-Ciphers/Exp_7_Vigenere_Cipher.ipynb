```python
# Exp_7_Vigenere_Cipher.py

def vigenere_cipher(text, key, mode='encrypt'):
    result = ""
    key = key.upper()
    key_length = len(key)
    
    for i, char in enumerate(text.upper()):
        if char.isalpha():
            # Get 0-25 value of the plaintext and the current key letter
            p = ord(char) - 65
            k = ord(key[i % key_length]) - 65
            
            if mode == 'encrypt':
                enc = (p + k) % 26
            else:
                enc = (p - k + 26) % 26 # +26 handles negative numbers in Python securely
                
            result += chr(enc + 65)
        else:
            result += char
    return result

plain_text = "ATTACKATDAWN"
keyword = "LEMON"
cipher = vigenere_cipher(plain_text, keyword, 'encrypt')

print(f"Vigenere Encrypt: {plain_text} (Key: {keyword}) -> {cipher}")
print(f"Vigenere Decrypt: {cipher} -> {vigenere_cipher(cipher, keyword, 'decrypt')}")
```
