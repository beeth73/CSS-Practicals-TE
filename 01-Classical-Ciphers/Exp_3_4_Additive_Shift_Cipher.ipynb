```python
# Exp_3_4_Additive_Shift_Cipher.ipynb

def shift_cipher(text, key, mode='encrypt'):
    result = ""
    # If decrypting, we just shift backwards
    if mode == 'decrypt':
        key = -key
        
    for char in text.upper():
        if char.isalpha():
            # 1. Convert to 0-25
            x = ord(char) - 65
            # 2. Add key and Modulo 26
            enc = (x + key) % 26
            # 3. Convert back to ASCII
            result += chr(enc + 65)
        else:
            result += char
    return result

plain_text = "FRCRCE"
key = 3
cipher = shift_cipher(plain_text, key, 'encrypt')
print(f"Additive/Shift Encrypt: {plain_text} -> {cipher}")
print(f"Additive/Shift Decrypt: {cipher} -> {shift_cipher(cipher, key, 'decrypt')}")
```
