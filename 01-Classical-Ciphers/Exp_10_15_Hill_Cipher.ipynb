```python
# Exp_10_15_Hill_Cipher.py

# A 2x2 Key Matrix
key_matrix = [
    [3, 3],
    [2, 5]
]

def hill_encrypt_2x2(text, matrix):
    # Pad with 'X' if length is odd to make it blocks of 2
    if len(text) % 2 != 0:
        text += 'X'
        
    cipher_text = ""
    
    # Process 2 letters at a time
    for i in range(0, len(text), 2):
        p1 = ord(text[i]) - 65
        p2 = ord(text[i+1]) - 65
        
        # Pure Python Matrix Multiplication! [P1, P2] * Matrix
        c1 = (p1 * matrix[0][0] + p2 * matrix[1][0]) % 26
        c2 = (p1 * matrix[0][1] + p2 * matrix[1][1]) % 26
        
        cipher_text += chr(c1 + 65) + chr(c2 + 65)
        
    return cipher_text

plain_text = "DIAL"
cipher = hill_encrypt_2x2(plain_text, key_matrix)
print(f"Hill Cipher Encrypt (2x2): {plain_text} -> {cipher}")
```
