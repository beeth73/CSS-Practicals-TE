```python
# Exp_6_Vernam_Cipher.py
import random

def generate_vernam_key(length):
    # Generates a random key of EXACTLY the message length
    return "".join(chr(random.randint(0, 25) + 65) for _ in range(length))

def vernam_cipher(text, key, mode='encrypt'):
    if len(text) != len(key):
        return "Error: Key length must match Plaintext length!"
        
    result = ""
    for p_char, k_char in zip(text.upper(), key.upper()):
        p = ord(p_char) - 65
        k = ord(k_char) - 65
        
        if mode == 'encrypt':
            enc = (p + k) % 26
        else:
            enc = (p - k + 26) % 26
            
        result += chr(enc + 65)
    return result

msg = "SECRET"
# Must generate a key of exactly length 6
otp_key = generate_vernam_key(len(msg)) 

cipher = vernam_cipher(msg, otp_key, 'encrypt')
print(f"Vernam Encrypt: {msg} + Key({otp_key}) -> {cipher}")
print(f"Vernam Decrypt: {cipher} -> {vernam_cipher(cipher, otp_key, 'decrypt')}")
```
