```python
# Exp_2_Salt_and_Pepper.py
import random
import string

# Dummy hash function to simulate MD5/SHA1 without importing libraries
def simple_dummy_hash(text):
    hash_val = 0
    for char in text:
        hash_val = (hash_val * 31 + ord(char)) & 0xFFFFFFFF
    return hex(hash_val)

def generate_salt(length=8):
    return "".join(random.choices(string.ascii_letters + string.digits, k=length))

# The Pepper is a global secret hidden in the server's backend code
SECRET_PEPPER = "FrCrce_Secret_Key_99!"

def register_user(password):
    salt = generate_salt()
    # COMBINE: Pepper + Password + Salt
    combined_string = SECRET_PEPPER + password + salt
    
    hashed_password = simple_dummy_hash(combined_string)
    
    print(f"--- USER REGISTRATION ---")
    print(f"Original Password : {password}")
    print(f"Generated Salt    : {salt}")
    print(f"String to Hash    : {combined_string}")
    print(f"Final DB Hash     : {hashed_password}")
    return salt, hashed_password

def login_user(input_password, stored_salt, stored_hash):
    # To verify, we reconstruct the exact same string
    test_string = SECRET_PEPPER + input_password + stored_salt
    test_hash = simple_dummy_hash(test_string)
    return test_hash == stored_hash

# Run the simulation
db_salt, db_hash = register_user("MySuperPassword123")

print("\n--- LOGIN ATTEMPT ---")
print("Login with correct password:", login_user("MySuperPassword123", db_salt, db_hash))
print("Login with wrong password:  ", login_user("WrongPass", db_salt, db_hash))
```
