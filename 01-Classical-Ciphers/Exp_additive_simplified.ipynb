```python
def shift_cipher(text, key, mode='encrypt'):
    result = ""
    if mode == 'decrypt':
        key = -key

    for char in text:
        if char.isalpha():
            base = ord('A') if char.isupper() else ord('a')
            result += chr((ord(char) - base + key) % 26 + base)
        else:
            result += char
    return result
```

```python
print(shift_cipher("Hello, World!", 3))
# Khoor, Zruog!
```
