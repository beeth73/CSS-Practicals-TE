```python
# Exp_1_Product_Cipher.py

# We will reuse the shift_cipher logic from above for the Substitution phase
def substitution_phase(text, key):
    return "".join(chr((ord(c) - 65 + key) % 26 + 65) for c in text.upper() if c.isalpha())

# Transposition Phase: We will use a simple reverse string for demonstration
def transposition_phase(text):
    return text[::-1] 

def product_encrypt(plain_text, key):
    subbed = substitution_phase(plain_text, key)
    transposed = transposition_phase(subbed)
    return transposed

def product_decrypt(cipher_text, key):
    un_transposed = transposition_phase(cipher_text) # Reversing a reverse fixes it
    un_subbed = substitution_phase(un_transposed, -key)
    return un_subbed

msg = "HELLO"
prod_cipher = product_encrypt(msg, 3)
print(f"Product Encrypt (Shift + Reverse): {msg} -> {prod_cipher}")
print(f"Product Decrypt: {prod_cipher} -> {product_decrypt(prod_cipher, 3)}")
```
