```python
# Exp_8_Playfair_Simplified.py

def generate_playfair_grid(key):
    # Remove duplicates and replace J with I
    key = "".join(dict.fromkeys(key.upper().replace("J", "I")))
    alphabet = "ABCDEFGHIKLMNOPQRSTUVWXYZ" # No J
    
    # Create the 25-letter flat list
    grid_str = key + "".join([c for c in alphabet if c not in key])
    
    # Convert to 5x5 matrix
    return [list(grid_str[i:i+5]) for i in range(0, 25, 5)]

def find_position(grid, letter):
    for r in range(5):
        for c in range(5):
            if grid[r][c] == letter:
                return r, c

def playfair_encrypt_simplified(text, key):
    grid = generate_playfair_grid(key)
    text = text.upper().replace("J", "I").replace(" ", "")
    
    # Ensure even length by padding with 'X'
    if len(text) % 2 != 0:
        text += 'X'
        
    cipher_text = ""
    
    for i in range(0, len(text), 2):
        r1, c1 = find_position(grid, text[i])
        r2, c2 = find_position(grid, text[i+1])
        
        if r1 == r2:
            # Rule 1: Same Row -> Shift Right (Wrap around using % 5)
            cipher_text += grid[r1][(c1 + 1) % 5] + grid[r2][(c2 + 1) % 5]
        elif c1 == c2:
            # Rule 2: Same Column -> Shift Down (Wrap around using % 5)
            cipher_text += grid[(r1 + 1) % 5][c1] + grid[(r2 + 1) % 5][c2]
        else:
            # Syllabus says "only first two rules", so we just leave as is if diagonal
            cipher_text += text[i] + text[i+1]
            
    return cipher_text

key = "KEYWORD"
plain_text = "BALL" 
cipher = playfair_encrypt_simplified(plain_text, key)

print(f"Playfair Grid Key: {key}")
for row in generate_playfair_grid(key): print(row)
print(f"\nSimplified Playfair Encrypt: {plain_text} -> {cipher}")
```
