```python
# Exp_5_Multiplicative_Cipher.py

# To decrypt, we need the Multiplicative Inverse of the key (key^-1)
# such that (key * inverse) % 26 == 1
def mod_inverse(key, modulo=26):
    for i in range(1, modulo):
        if (key * i) % modulo == 1:
            return i
    return -1 # Fails if key is not coprime to 26

def multiplicative_cipher(text, key, mode='encrypt'):
    result = ""
    if mode == 'decrypt':
        key = mod_inverse(key)
        if key == -1:
            return "Invalid key! Key must be coprime to 26."

    for char in text.upper():
        if char.isalpha():
            x = ord(char) - 65
            enc = (x * key) % 26
            result += chr(enc + 65)
        else:
            result += char
    return result

plain_text = "SECURITY"
key = 7
cipher = multiplicative_cipher(plain_text, key, 'encrypt')
print(f"Multiplicative Encrypt: {plain_text} -> {cipher}")
print(f"Multiplicative Decrypt: {cipher} -> {multiplicative_cipher(cipher, key, 'decrypt')}")
```
