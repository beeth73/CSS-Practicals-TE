```python
# Exp_16_17_Simple_Columnar.py

def columnar_transposition(text, num_cols):
    # Clean the text
    text = text.replace(" ", "").upper()
    
    # 1. Pad the text with 'X' so it perfectly fits the columns
    while len(text) % num_cols != 0:
        text += 'X'
        
    # 2. Create empty strings for each column
    columns = ["" for _ in range(num_cols)]
    
    # 3. Write Row-by-Row (Simulated by sending each char to the correct column)
    for i, char in enumerate(text):
        columns[i % num_cols] += char
        
    # 4. Read Column-by-Column to get the cipher text
    cipher_text = "".join(columns)
    
    return cipher_text

plain_text = "WE ARE DISCOVERED FLEE AT ONCE"

# Experiment 17: 3 Columns
cipher_3 = columnar_transposition(plain_text, 3)
print(f"3-Column Cipher: {cipher_3}")

# Experiment 16: 4 Columns
cipher_4 = columnar_transposition(plain_text, 4)
print(f"4-Column Cipher: {cipher_4}")
```
