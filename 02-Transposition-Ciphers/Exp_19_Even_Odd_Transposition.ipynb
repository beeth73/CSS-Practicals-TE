```python
# Exp_19_Even_Odd_Transposition.py

def even_odd_transposition(text):
    # Remove spaces and capitalize
    text = text.replace(" ", "").upper()
    
    # Python slicing: text[start:stop:step]
    # Human EVEN positions are Python indices 1, 3, 5...
    even_chars = text[1::2]
    
    # Human ODD positions are Python indices 0, 2, 4...
    odd_chars = text[0::2]
    
    # Put Evens BEFORE Odds as requested by the syllabus
    return even_chars + odd_chars

# Test with the exact example from the syllabus!
plain_text = "EXAM"
cipher = even_odd_transposition(plain_text)

print(f"Plaintext : {plain_text}")
print(f"Ciphertext: {cipher}")
print(f"Match syllabus example? {cipher == 'XMEA'}")
```
